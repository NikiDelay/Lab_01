```python
# ============================================================
# Задание 3: Сумма факториалов
# ============================================================
import math

def task_03_01_func(n):
    sum = 0  
    for i in range(n + 1):
        sum = sum + math.factorial(i)
    return sum

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

def task_03_02_func(n):
    if n == 0:
        return 1 
    else:
        return task_03_02_func(n - 1) + factorial(n)
    
def test_03_func():
    assert task_03_01_func(4) == 34  
    assert task_03_01_func(13) == 6749977114
    assert task_03_02_func(5) == 154
    assert task_03_02_func(10) == 4037914 
    assert task_03_02_func(0) == 1 
    print("Все тесты пройдены!")

test_03_func()
```

```python
# ============================================================
# Задание 4: Проверка палиндрома 
# ============================================================
def task_04_01_func(s):
    return s == s[::-1]

def task_04_02_func(s):
    left = 0
    right = len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

def task_04_03_func(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return task_04_03_func(s[1:-1])
def test_04_func():
    assert task_04_01_func('казак') == True
    assert task_04_01_func('шалаш') == True
    assert task_04_02_func('привет') == False
    assert task_04_02_func('А') == True
    assert task_04_03_func('') == True
    assert task_04_03_func('abba') == True
    print("Все тесты пройдены!")

test_04_func()
```

```python
# ============================================================
# Задание 5: Частота слов 
# ============================================================
def task_05_01_func(text):
    words = text.split()
    result = {}
    
    for word in words:
        if word in result:
            result[word] += 1
        else:
            result[word] = 1
    
    return result

def task_05_02_func(text):
    words = text.split()
    result = {}
    
    for word in words:
        result[word] = result.get(word, 0) + 1
    
    return result

def task_05_03_func(text):
    words = text.split()
    result = {}
    
    for word in words:
        result.setdefault(word, 0)
        result[word] += 1
    
    return result

def test_05_func():
    assert task_05_01_func("hello world hello") == {'hello': 2, 'world': 1}
    assert task_05_01_func("") == {}
    assert task_05_02_func("Ты поднял меч") == {'Ты': 1, 'поднял': 1, 'меч': 1}
    assert task_05_02_func("a a a") == {'a': 3}
    assert task_05_03_func("test test test test test") == {'test': 5}
    assert task_05_03_func("123 456 123") == {'123': 2, '456': 1}
    print("Все тесты пройдены!")
test_05_func()
```

```python
# ============================================================
# Задание 9: Треугольник Паскаля 
# ============================================================
def task_09_func(n):
    rows = []
    row = [1]
    for i in range(n):
        rows.append(row)
        new_row = [1]
        for j in range(len(row) - 1):
            new_row.append(row[j] + row[j + 1])
        new_row.append(1)
        row = new_row
    
    max_num = max(max(r) for r in rows)
    cell_width = len(str(max_num)) +2
    
    for i, row in enumerate(rows):
        formatted = ''.join(str(x).center(cell_width) for x in row)
        indent = ' ' * (cell_width * (n - i - 1) // 2)
        print(indent + formatted)

number = int(input())
task_09_func(number)
```

```python
# ============================================================
# Задание 10: Переименование файлов 
# ============================================================
from pathlib import Path

def task_10_func(dir_path, prev_extension, next_extension):
    directory = Path(dir_path)
    
    if not directory.exists() or not directory.is_dir():
        return (0, 0)
    
    total_files = 0
    changed_count = 0
    
    for item in directory.iterdir():
        if item.is_file():
            total_files += 1
            
            if item.suffix == prev_extension:
                new_path = item.with_suffix(next_extension)
                
                if new_path.exists():
                    continue
                
                item.rename(new_path)
                changed_count += 1
                print(f"{item.name} => {new_path.name}")
                
    print(f"Всего файлов: {total_files}")
    print(f"Переименовано: {changed_count}")
    
    return (total_files, changed_count)

result = task_10_func("C:/Users/user/Desktop/dir", ".bak", ".txt")
```

```python
# ============================================================
# Задание 12: Обработка файла с числами (1 балл)
# ============================================================
def process_numbers(input_path, output_path):
    # ПРОХОД 1
    min_val = float('inf')
    
    with open(input_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            
            num = float(line)
            if num < min_val:
                min_val = num

    # ПРОХОД 2
    with open(input_path, 'r', encoding='utf-8') as f_in, \
         open(output_path, 'w', encoding='utf-8') as f_out:
        
        for i, line in enumerate(f_in):
            line = line.strip()
            if not line:
                continue
                
            if i % 2 == 0:
                num = float(line)
                result = num + min_val
                f_out.write(f"{result:.5f}\n")
                print(f"Строка {i}: {num} + {min_val} = {result:.5f}")
process_numbers('C:/Users/user/Desktop/file12/input.txt', 'C:/Users/user/Desktop/file12/output.txt')
```

```python
# ============================================================
# Задание 13: Генератор чисел Фибоначчи (1 балл)
# ============================================================
def task_13_func(n):
    a, b = 0, 1
    if n == -1:
        while True:
            yield a
            a, b = b, a + b
    else:
        for _ in range(n):
            yield a
            a, b = b, a + b

n = int(input())
if n == -1:
    print('♾️')
else:
    print(list(task_13_func(n))) 
```

```python
# ============================================================
# Задание 14: Магические методы объекта (1 балл)
# ============================================================
def task_14_func(obj):
    methods = dir(obj)
    
    def is_magic_a(m):
        return m.startswith('__a') and m.endswith('__')
    
    def is_magic_s(m):
        return m.startswith('__s') and m.endswith('__')
    
    def is_not_magic(m):
        return not (m.startswith('__') and m.endswith('__'))
    
    if isinstance(obj, int) and not isinstance(obj, bool):
        return list(filter(is_magic_a, methods))
    elif isinstance(obj, str):
        return list(filter(is_magic_s, methods))
    else:
        return list(filter(is_not_magic, methods))

def test_14_func():
    # int — магические методы на '__a'
    result_int = task_14_func(5)
    assert isinstance(result_int, list)
    assert '__abs__' in result_int
    assert '__add__' in result_int
    assert '__and__' in result_int
    assert all(m.startswith('__a') and m.endswith('__') for m in result_int)
    # str — магические методы на '__s'
    result_str = task_14_func("hello")
    assert isinstance(result_str, list)
    assert '__str__' in result_str
    assert '__sizeof__' in result_str
    assert all(m.startswith('__s') and m.endswith('__') for m in result_str)
    # list — немагические методы
    assert task_14_func([1, 2, 3]) == [
        'append', 'clear', 'copy', 'count', 'extend',
        'index', 'insert', 'pop', 'remove', 'reverse', 'sort'
    ]
    # tuple — немагические методы
    assert task_14_func((1, 2)) == ['count', 'index']
    # bool — должен попасть в "другие типы" (немагические)
    result_bool = task_14_func(True)
    assert all(not (m.startswith('__') and m.endswith('__')) for m in result_bool)
    result_empty = task_14_func("")
    assert '__str__' in result_empty
    assert all(m.startswith('__s') and m.endswith('__') for m in result_empty)

    print("Все тесты пройдены!")

test_14_func()
```

```python
# ============================================================
# Задание 15
# ============================================================
import re


def _replace_upper(match):
    return str(ord(match.group()))


def task_15_func_01(s):

    # Заменяем заглавные буквы на их ASCII-коды
    result = re.sub(r'[A-Z]', _replace_upper, s)

    # Находим все цифры в результате
    digits = list(filter(str.isdigit, result))
    if not digits:
        return [result]

    # Находим минимальную цифру и сплитим по ней
    min_digit = min(digits)
    return result.split(min_digit)


def task_15_func_02(s):

    result = []
    digits = []

    # Первый проход: заменяем буквы и собираем цифры
    for char in s:
        if 'A' <= char <= 'Z':
            ascii_code = str(ord(char))
            result.append(ascii_code)
            for digit in ascii_code:
                digits.append(digit)
        else:
            result.append(char)
            if char.isdigit():
                digits.append(char)

    result_str = ''.join(result)

    # Второй проход: находим минимум и сплитим
    if not digits:
        return [result_str]

    min_digit = min(digits)
    return result_str.split(min_digit)


# Тесты
test_string = "Hello ABC"
print(f"Входная строка: {test_string}")
print(f"Результат 01: {task_15_func_01(test_string)}")
print(f"Результат 02: {task_15_func_02(test_string)}")
print()

test_cases = [
    "Hello ABC",
    "Test XYZ",
    "abc",
    "A1B2",
]

for test in test_cases:
    print(f"'{test}' → v1: {task_15_func_01(test)}, v2: {task_15_func_02(test)}")
```

```python
# ============================================================
# Задание 16: Извлечение телефонных номеров 
# ============================================================
import re

def task_16_func(n):

    pattern = r'(?:\+7|8)[\s\-]?\(?\d{3}\)?[\s\-]?(?:\d{3}[\s\-]?\d{2}[\s\-]?\d{2}|\d{7})'

    matches = re.findall(pattern, n)

    result = []
    for match in matches:
        # Убираем все нецифровые символы
        digits = re.sub(r'\D', '', match)

        # Нормализация
        if digits.startswith('8') and len(digits) == 11:
            digits = '7' + digits[1:]

        # Добавляем + и сохраняем только валидные 11-значные номера
        if len(digits) == 11:
            result.append('+' + digits)

    return result


if __name__ == "__main__":
    # Тестовая строка со всеми 10 форматами
    text = """
    Мой номер: +7 901 121-17-43
    Другой: 8-901-121-17-43
    Ещё: +7(901)121-17-43
    Без разделителей: 89011211743
    С пробелами: 8 901 121 17 43
    Смешанный: +7 901 1211743
    С дефисами: +7-901-121-17-43
    Со скобками: 8 (901) 121-17-43
    Комбо: 8-901-1211743
    Плюс: +79011211743
    """

    result = task_16_func(text)
    print("Найденные номера:")
    for i, num in enumerate(result, 1):
        print(f"  {i}. {num}")
    print(f"\nВсего найдено: {len(result)} номеров")
```
